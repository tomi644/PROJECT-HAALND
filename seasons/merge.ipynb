```python
import pandas as pd
from pathlib import Path
```

```python
folder = Path.cwd()

print("Current folder:", folder)

files = sorted(folder.glob("*.xls"))

print("Number of Excel files:", len(files))

for file in files:
    print(file.name)
```

```python
import pandas as pd

df = pd.read_excel('2023-24.xls', engine='xlrd')

df.head()
```

```python
season_data = []

for file in files:
    df = pd.read_excel(file, engine="xlrd")

    df["Season"] = file.stem

    season_data.append(df)

combined = pd.concat(season_data, ignore_index=True)

combined.head()
```
