```python

import pandas as pd
from pathlib import Path

# Folder containing this Python script
folder = Path(__file__).parent

# Find all Excel .xls files, excluding temporary files
files = [
    file for file in folder.glob("*.xls")
    if not file.name.startswith("~$")
]

dataframes = []

for file in files:
    print(f"Reading {file.name}...")

    df = pd.read_excel(file, engine="xlrd")

    # Keep track of which file each row came from
    df["source_file"] = file.name

    dataframes.append(df)

# Combine all files vertically
combined = pd.concat(dataframes, ignore_index=True)

# Save the result
output_file = folder / "combined.xlsx"
combined.to_excel(output_file, index=False)

print(f"Done! Combined {len(files)} files.")
print(f"Output saved to: {output_file}")
```
